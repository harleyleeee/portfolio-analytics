```python
import urllib.request as req
from bs4 import BeautifulSoup
import pandas as pd 
from time import sleep 
from random import randint
```

```python
url = "https://www.freelancer.com/freelancers/canada/python/1"
html = req.urlopen(url)
doc = BeautifulSoup(html, "html.parser")
```

```python
name0= doc.find_all(class_="find-freelancer-username-mobile")
name0
```

```python
len(doc.find_all(class_="find-freelancer-username-mobile"))
```

```python
for n in name0:
    print(n.text)
```

```python
name1 = [n.text.strip() for n in name0]
name1
```

```python
review0=doc.find_all(class_="Rating Rating--labeled")
review0
```

```python
review1 = [r["data-star_rating"] for r in review0]
review1
```

```python
earning0= doc.find_all(class_= "Earnings-label freelancer-user-earnings")
earning0
```

```python
for e in earning0:
    print(e.text)
```

```python
earning1= [e.text.strip() for e in earning0]
earning1
```

```python
number0 = doc.find_all(class_="directory-freelancer-rating")
number0
```

```python
for l in number0:
   print(l.text)
```

```python
number1=[l.text.strip() for l in number0]
number1
```

```python
len(number1)
```

```python
rate0 =doc.find_all(class_="user-hourly-rate freelancer-hourlyrate")
rate0
```

```python
for r in rate0:
    print(r.text)
```

```python
rate1=[r.text for r in rate0]
rate1
```

```python
len(rate1)
```

```python
profile0=doc.find_all(class_="bio truncProfile")
profile0
```

```python
len(profile0)
```
