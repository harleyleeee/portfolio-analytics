# 1. Crawl data using API of TMDB (Themoviedb.org)

```python
import requests
import pandas as pd

API_KEY = "8ebc019c041f4ba9fdc43675eb228764"
BASE_URL = "https://api.themoviedb.org/3/discover/movie"
MOVIE_URL = "https://api.themoviedb.org/3/movie/"

movies = []
page = 1
total_pages = 30 

while page <= total_pages:
    params = {
        "api_key": API_KEY,
        "language": "en-US",
        "sort_by": "popularity.desc",
        "primary_release_date.gte": "2025-01-01",
        "primary_release_date.lte": "2025-06-01",
        "page": page,
    }
    #Send request to API
    response = requests.get(BASE_URL, params=params)
    data = response.json()
    
    print(f"Page {page} - Movies found: {len(data.get('results', []))}")
    
    for movie in data.get("results", []):
        # Get information about each movie
        movie_id = movie["id"]
        movie_details = requests.get(f"{MOVIE_URL}{movie_id}", params={"api_key": API_KEY, "language": "en-US"}).json()
    
        box_office_earnings = movie_details.get("revenue", 0)  # Box office earnings 
        runtime = movie_details.get("runtime", 0)  # Runtime
        
        movies.append({
            "id": movie["id"],
            "title": movie["title"],
            "original_title": movie.get("original_title", ""),
            "release_date": movie.get("release_date", ""),
            "popularity": movie.get("popularity", 0),
            "vote_average": movie.get("vote_average", 0),
            "vote_count": movie.get("vote_count", 0),
            "overview": movie.get("overview", ""),
            "runtime": runtime,
            "box_office_earnings": box_office_earnings
        })
    
    page += 1
```

```python
df_movies = pd.DataFrame(movies)
df_movies
```

### Convert the variable "release_date" which is under the form of date into the numeric variables named "Days_since_release"

```python
df_movies['release_date'] = pd.to_datetime(df_movies['release_date'])
```

```python
reference_date = pd.to_datetime('2025-06-13')
```

```python
df_movies['days_since_release'] = (reference_date-df_movies['release_date']).dt.days
df_movies
```

# 2. Sentiment Analysis of Film Overview

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
```

```python
import nltk
nltk.download('vader_lexicon')
```

```python
vader = SentimentIntensityAnalyzer()
```

```python
df_movies["overview"]
```

```python
overview_list = df_movies["overview"].tolist()
```

```python
overview_list
```

```python
overview_list[0]
```

```python
vader.polarity_scores(overview_list[0])
```

```python
vader.polarity_scores(overview_list[0])["compound"]
```

```python
sentiments = [vader.polarity_scores(o)["compound"] for o in overview_list]
```

```python
sentiments
```

```python
df_movies["Overview_sentiment"] = sentiments
```
