# 4. How do transaction patterns vary across different age groups and genders?

```python
df.head(5)
```

Define the age group

```python
def phan_loai_tuoi(tuoiKH):
  if tuoiKH >= 66:
    return 'Senior(66+)'
  elif 51<= tuoiKH <= 65:
    return 'Baby Boomers(51-65)'
  elif 26 <=tuoiKH <= 35:
    return 'Millennials(26-35)'
  elif 36 <= tuoiKH <= 50:
    return 'Gen X(36-50)'
  else:
    return 'Gen Z(18-25)'
```

```python
df["NhomTuoi"] = df["Tuoi"].apply(phan_loai_tuoi)
df.head(5)
```

```python
#Number of transactions for each age group
SoLuongGD_NhomTuoi=df.groupby("NhomTuoi")["MaGD"].count()

#Total volume of transactions for each age group
TongTien_NhomTuoi =df.groupby("NhomTuoi")["SoTien"].sum()

#Average volume of transaction for each age group
GDTB_NhomTuoi = df.groupby("NhomTuoi")["SoTien"].mean()

#Number of customer for each age group
SoLuongKH_NhomTuoi = df.groupby("NhomTuoi")["MaKH"].nunique()
```

```python
#Transaction patterns across different age groups
df_tonghop1= pd.DataFrame({
    "Số GD": SoLuongGD_NhomTuoi,
    "Tổng GD (Tỷ)": TongTien_NhomTuoi/1e9,
    "GD TB (TR)": GDTB_NhomTuoi/1e6,
    "Số KH": SoLuongKH_NhomTuoi
})
df_tonghop1
```

```python
fig,ax3=plt.subplots(figsize=(12,6))
ax3.set_title("Tổng Giá Trị GD theo Nhóm Tuổi")
ax3.set_xlabel("Nhóm Tuổi")
ax3.set_ylabel("Tỷ VNĐ")

ax3.bar_label(
    bars,
    labels=[f"{x:.1f}" for x in df_tonghop1["Tổng GD (Tỷ)"].values],
    padding=3
)
colors = plt.cm.viridis([i/len(df_tonghop1) for i in range(len(df_tonghop1))])
bars = ax3.bar(df_tonghop1.index,df_tonghop1["Tổng GD (Tỷ)"].values, color = colors)

plt.tight_layout()
plt.show()
```

Conclusion: Baby boommers contributes the highest number and largest volume of transactions among 5 provided age groups

# 5. How does transaction volume by channel vary across different time periods of the day?

```python
df["OnlyGioGD"] = df["GioGD"].str.split(":").str[0].astype(int)
df["OnlyGioGD"]
```

```python
pivot_table3
```

```python
#Visualization: heatmap
fig,ax8=plt.subplots(figsize=(12,6))
sns.heatmap(pivot_table3, cmap="YlGnBu", annot=True, fmt=".0f", ax=ax8)
ax8.set_title("Heatmap: Số GD Theo Kênh x Giờ Trong Ngày")
ax8.set_xlabel("Giờ GD")
ax8.set_ylabel("Kênh GD")

plt.tight_layout()
plt.show()
```

Conclusion:
1. For ATM, 2pm is the peak time for transactions
2. For Internet Banking, 4pm is the peak time for transactions
3. For Mobile App, 3pm is the peak time for transactions
4. For POS, 1pm is the peak time for transactions
5. For offline-store, 9am is the peak time for transactions

# 6. Who are the top 20 customers by total transaction value?

```python
Top20 = df.sort_values("SoTien", ascending=False).head(20)
Top20
```

```python
Top20["HoTen"].values
```

```python
#Visualization: barth
from matplotlib.patches import Patch
fig,ax9=plt.subplots(figsize=(12,6))


nhom = Top20["NhomKH"].unique()

color_map = {n: plt.cm.tab10(i) for i, n in enumerate(nhom)}

colors= Top20["NhomKH"].map(color_map)
bars = ax9.barh(
    Top20["HoTen"]+" (" + Top20["MaKH"] + ")",
    Top20["SoTien"].values/1e9,
    color=colors)

ax9.bar_label(
    bars,
    labels =[f"{x/1e9:.1f}" for x in Top20["SoTien"].values],
    padding=3
)
ax9.invert_yaxis()

ax9.set_title("Top20 Khách hàng Giá trị nhất")

legend = [
    Patch(color=color, label=nhom)
    for nhom, color in color_map.items()
]

ax9.legend(
    handles=legend,
    title="Nhóm KH",
    loc="lower right"
)

plt.tight_layout()
plt.show()
```

Conclusion: Top 1 customer by total transaction values goes to VIP individual types
